# **Dictionary in python**

# create Dictionary

```python
# 17.1) create Dictionary
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person)
print(type(person))
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person['name'])
print(person['age'])
print(person['address'])
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person.get('name'))
print(person.get('age'))
print(person.get('address'))
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person.keys())
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person.values())
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
print(person.items())
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
for key in person.keys():
    print(key)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
for value in person.values():
    print(value)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
for key,value in person.items():
    print(key,value)
```

# dict () constructor

```python
# dict () constructor
person = dict(name='meesal',age=33,address='shahabad')
print(person)
```

```python
person = dict(name='meesal',age=33,address='shahabad')
for i in person.items():
    print(i)
```

# check if key exist

```python
# check if key exist
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
if 33 in person.values():
    print('yes')
else:
    print('no')
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
if 'meesal' in person.values():
    print('yes')
else:
    print('no')
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
if 'apple' in person.values():
    print('yes')
else:
    print('no')
```

# change values

```python
# change values
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person['name'] = 'yanaguntikar'
print(person)
```

# update ()

```python
# update ()
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person.update({'name':'yanaguntikar'})
print(person)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person.update({'mobile':124578})
print(person)
```

# adding items

```python
# adding items
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person['mobile'] = 124578
print(person)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person['EMP NO'] = 4578
print(person)
```

# removing (pop) items

```python
# removing (pop) items
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person.pop('name')
print(person)
```

# Del () method

```python
# Del () method
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
del person['name']
print(person)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
del person
print(person)
```

# clear () method

```python
# clear () method
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
person.clear()
print(person)
```

# copy () method

```python
# copy () method
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
new_person = person.copy()
print(new_person)
```

# Nested Dictionaries

```python
# Nested Dictionaries
person = {

    'student_1':{
        'name':'meesal',
        'age':33,
        'address':'shahabad'
    },
    'student_2':{
        'name':'yanaguntikar',
        'age':25,
        'address':'kalaburgi'
    },
    'student_3':{
        'name':'amar',
        'age':39,
        'address':'bangolre'
}

}
print(person)
```

```python
person = {

    'student_1':{
        'name':'meesal',
        'age':33,
        'address':'shahabad'
    },
    'student_2':{
        'name':'yanaguntikar',
        'age':25,
        'address':'kalaburgi'
    },
    'student_3':{
        'name':'amar',
        'age':39,
        'address':'bangolre'
}

}
for i in person.items():
    print(i)
```

# form keys () method

```python
# form keys () method
person = ('key1','key2','key3')
new_person = dict.fromkeys(person)
print(new_person)
```

# set default () method

```python
# set default () method
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
new_person = person.setdefault('name','yanaguntikar')
print(new_person)
```

```python
person = {
    'name':'meesal',
    'age':33,
    'address':'shahabad'
}
new_person = person.setdefault('mobile',124578)
print(new_person)
```
