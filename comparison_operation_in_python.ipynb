# comparison operation in python

# 1. Equal to (==)

```python
# 4.1) Equal to (==)
a = 5
b = 5
print(a == b)
```

# 2. Not equal to (!=)

```python
# 4.2) Not equal to (!=)
a = 5
b = 10
print(a != b)
```

# 3. Greater than (>)

```python
# 4.3) Greater than (>)
a = 10
b = 5
print(a > b)
```

# 4. Less than (< )

```python
# 4.4) Less than (< )
a = 5
b = 10
print(a < b)
```

# 5. Greater than or equal to ( >= )

```python
# 4.5) Greater than or equal to ( >= )
a = 10
b = 10
print(a >= b)
```

# 6. Less than or equal to ( <= )

```python
# 4.6) Less than or equal to ( <= )
a = 5
b = 10
print(a <= b)
```

# 7. comparing string

```python
# 4.7) comparing string
a = "apple"
b = "banana"
print(a < b)
```

# 8. comparing lists

```python
# 4.8) comparing lists
a = [1,2,3]
b = [1,2,3]
print(a == b)
```
