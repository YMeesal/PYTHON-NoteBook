# separate positive and negative integers from the tuple

---



```python
# 12.19) using loops in tupel
# separate positive and negative integers from the tuple
a = (1,2,3,-4,-5,-6)
pos = []
neg = []
a = list(a)
for i in range(len(a)):
  if a[i] >= 0:
    pos.append(a[i])
  else:
    neg.append(a[i])
print("positive number : ",pos)
print("negative number : ",neg)
```
