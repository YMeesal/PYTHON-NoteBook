# Arithmetic operations in python

# Addition (+)

```python
# 1) Addition (+)
a = 5
b = 3
print("Result = ",a + b)
```

# Subtraction (-)

```python
# 3.2) Subtraction (-)
a = 10
b = 5
print("Result = ", a - b)
```

# Multiplication (*)

```python
# 3.3) Multiplication (*)
a = 5
b = 2
print("Result = ", a * b)
```

# Division (/)

```python
# 3.4) Division (/)
a = 10
b = 2
print("Result = ", a / b)
```

# Modulus (%)

```python
# 3.5) Modulus (%)
a = 10
b = 2
print("Result = ", a % b)
```

# Exponentiation (**)

---



```python
# 3.6) Exponentiation (**)
a = 10
b = 2
print("Result = ", a ** b)
```

# Floor Division (//)

```python
# 3.7) Floor Division (//)
a = 10
b = 2
print("Result = ", a // b)
```
