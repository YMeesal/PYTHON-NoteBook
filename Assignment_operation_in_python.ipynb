# **ASSIGNMENT OPERATIONS IN PYTHON**

# simple Assignment (=)

```python
a = 5
print(a)
```

# Addition Assignment (+=)

```python
a = 10
a += 5
print(a)
```

# Subtraction Assignment (-=)

```python
a = 10
a -= 5
print(a)
```

# Multiplication Assignments (*=)

```python
a = 5
a *=2
print(a)
```

# Division Assignments (/=)

```python
a = 10
a /= 2
print(a)
```

# Modulus Assignments ( %= )

```python
a = 10
a %= 2
print(a)
```

# Exponentiation Assignment ( **= )

```python
a = 10
a **= 2
print(a)
```

# Floor Division Assignment ( //= )

```python
a = 10
a //= 2
print(a)
```

# Bitwise And Assignments ( &= )

```python
a = 5
a &= 2
print(a)
```

# Bitwise OR Assignments (|= )

```python
a = 5
a |= 3
print(a)
```

# Bitwise XOR Assignments ( ^= )

```python
a = 5
a ^= 3
print(a)
```

# Bitwise Left shift Assignments ( <<= )

```python
a = 5
a <<= 1
print(a)
```
