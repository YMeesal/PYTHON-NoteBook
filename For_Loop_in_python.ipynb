# **For Loop in python**

# 1. simple for loop to iterate over a list

```python
# 10.1) simple for loop to iterate over a list
fruits = ['apple','banana','cherry','mango']
for fruit in fruits:
    print(fruit)
```

# 2. for loop to iterate over a string

```python
# 10.2) for loop to iterate over a string
s = 'hello'
for char in s:
    print(char)
```

# 3. for loop to iterate over a range of numbers

```python
# 10.3) for loop to iterate over a range of numbers
for i in range(5):
    print(i)
```

# 4. for loop to iterate over a dictionary

```python
# 10.4) for loop to iterate over a dictionary
person = {'name':'john','age':30,'city':'newyork'}
for key,value in person.items():
    print(key,value)
```

# 5. for loop to iterate over a list of tuples

```python
# 10.5) for loop to iterate over a list of tuples
pairs = [(1,2), (3,4),(5,6)]
for a,b in pairs:
  print(a,b)
```

# 6. for loop with  an else clause

```python
# 10.6) for loop with  an else clause
for i in range(5):
  print(i)
else:
  print("loop finished")
```

# 7. for loop with a continue statement

```python
# 10.7) for loop with a continue statement
for i in range(10):
    if i == 5:
        continue  # Skip the iteration when i is 5
    print(i)
```
