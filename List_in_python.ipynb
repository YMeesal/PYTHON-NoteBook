# **List in python**

```python
# 11.1) create a list
a = [1,2,3,4,5]
print(a)
print(type(a))
```

# indexing on list

```python
# 11.2) indexing on list
a = [10,20,30,40,50,60,70,80]
print(a[0])
print(a[-1])
```

# slicing

```python
# 11.3) slicing
a = [10,20,30,40,50,60,70,80,90]
print(a[0:3])
print(a[::-1])
print(a[::2])
```

# Append method

```python
# 11.4) Append method
a = [1,2,3]
b = [4,5,6]
c = a + b
print(c)
```

```python
# 11.4.1) Append method
a = [1,2,3]
a.append(4)
print(a)
```

```python
# 11.4.2) Append method
a = ['apple','banana','cherry']
a.append('orange')
print(a)
```

# Extend () method

```python
# 11.5 ) Extend () method
num = [1,2,3]
num.extend([4,5,6])
print(num)
```

```python
# 11.5.1 ) Extend () method
fruits = ['apple','banana','cherry']
fruits.extend(['orange','kiwi'])
print(fruits)
```

# insert () method

```python
# 11.6 ) insert () method
num = [10,20,30,40,50]
num.insert(2,25)
print(num)
```

# Remove () method

```python
# 11.7) Remove () method
num = [10,20,30,40,50,60,70]
num.remove(50)
print(num)
```

# pop () method

```python
# 11.8) pop () method
num = [10,20,30,40,50]
num.pop()
print(num)
```

```python
# 11.8.1) pop () method
num = [10,20,30,40,50]
num.pop(1)
print(num)
```

```python
# 11.8.2) pop () method
num = [10,20,30,40,50]
num.pop(0)
print(num)
```

# clear () method

```python
# 11.9) clear () method
num = [10,20,30,40,50]
num.clear()
print(num)
```

```python
# 11.9.1) clear () method
fruits = ['apple','banana','cherry']
fruits.clear()
print(fruits)
```

# count () method

```python
# 11.10) count () method
num = [10,10,20,30,]
print(num.count(10))
```

# max () method

```python
# 11.11) max () method
num = [10,20,30,40,50]
print(max(num))
```

# min () method

```python
# 11.12) min () method
num = [10,20,30,40,50]
print(min(num))
```

# sort () method

```python
# 11.13) sort () method
num = [10,60,20,70,30,80,40,90,50]
num.sort()
print(num)
```

# reverse () method

```python
# 11.14) reverse () method
num = [10,20,30,40,50,60]
num.reverse()
print(num)
```

# index method

```python
# 11.15) index method
fruits = ['apple','banana','cherry']
print(fruits.index('banana'))
```

# copy () method

```python
# 11.16) copy () method
num = [10,20,30,40,50]
n = num.copy()
print(n)
```
