# CSV FILE FORMAT

```python
# CSV FILE FORMAT

import csv
data = [
    ['name','age','address'],
    ['meesal',23,'india'],
    ['yanaguntikar',33,'usa'],
    ['amar',21,'china'],
    ['rahul',46,'pakistan'],
    ['shetty',55,'uk']
]
with open("data.csv","w") as f:
    writer = csv.writer(f)
    writer.writerows(data)
```

# excel file format

```python
# excel file format

import pandas as pd

data_for_excel = [
    ['name','age','address'],
    ['meesal',23,'india'],
    ['yanaguntikar',33,'usa'],
    ['amar',21,'china'],
    ['rahul',46,'pakistan'],
    ['shetty',55,'uk']
]

# Create a pandas DataFrame from the list of lists, treating the first sublist as headers
df_excel = pd.DataFrame(data_for_excel[1:], columns=data_for_excel[0])

# Define the Excel file name with the correct extension
excel_file_name = "df.xlsx"

# Write the DataFrame to an Excel file without including the DataFrame index
df_excel.to_excel(excel_file_name, index=False)
print(f"Data successfully written to {excel_file_name}")
```

# 1. reading a file




```python
# 1. reading a file

with open("example.txt","w") as file:
  file.write("This is some example content.\n")
  file.write("Second line of content.")

with open("example.txt","r") as file:
  content = file.read()
  print(content)
```

# 2. writing to a file

```python
# 2. writing to a file

with open ("good.txt","w") as file:
  file.write("good morning\n")
  file.write("good night\n")
  file.write("good afternoon\n")
  file.write("good evening\n")
  file.write("good bye\n")

with open("good.txt","r") as file:
  hi = file.read()
  print(hi)
```

# Appending to a file

```python
with open("good.txt","a") as file:
  file.write("i am learning file handlig")
```

# reading a file line by line

```python
with open("good.txt","r") as file:
  for line in file:
    print(line.strip())
```

# writing multiple lines to a file

```python
lines = ['line 1','line 2','line 3','line 4']
with open("good.txt","w") as file:
  file.writelines("\n".join(lines))
```

```python
lines = ['line 1','line 2','line 3','line 4']
with open("good.txt","w") as file:
  for line in lines:
    file.write(line + "\n")
    print(line)
```

# checking if a file exists

```python
# checking if a file exists

import os
if os.path.exists("good.txt"):
  print("file exists")
else:
  print("file does not exist")
```

# Deleting  a file

```python
# Deleting  a file

import os
if os.path.exists("good.txt"):
  os.remove("good.txt")
  print("file deleted")
```
