# **Logical operation in python**

# AND

```python
# 7.1)  AND
a = 5
b = 10
print(a > 3 and b > 7)
```

# OR

```python
# 7.2) OR
name = "meesal"
age = 22
print(name == "meesal" or age > 20)
```

# NOT

```python
# 7.3) NOT
is_admin = True
print(not is_admin)
```

# AND

```python
# 7.4)  AND
fruits = ['apple','banana','cherry']
print('apple' in fruits and 'banana' in fruits)
```

# OR

```python
# 7.5) OR
person = {'name':'meesal','age':22}
print('name' in person or 'age' in person)
```

```python
person = {'name':'meesal','age':22}
print(person['name'] == 'meesal' or person['age'] > 20)
```

# NOT

```python
# 7.7) NOT
a = None
print(not a)
```
