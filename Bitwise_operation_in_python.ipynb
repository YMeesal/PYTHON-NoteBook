# **Bitwise operators in python**

# 1. Bitwise And ( & )

```python
# 5.1) Bitwise And ( & )
a = 5 # 101
b = 3 # 011
print(a & b)
```

# 2. Bitwise OR (|)

```python
# 5.2) Bitwise OR (|)
a = 5 # 101
b = 3 # 011
print(a | b)
```

# 3. Bitwise XOR (^)

```python
# 5.3) Bitwise XOR (^)
a = 5
b = 3
print(a ^ b)
```

# 4. Bitwise NOT (~)

```python
# 5.4) Bitwise NOT (~)
a = 5 # 101
print(~a)
```

# 5. Left shift (<<)

```python
# 5.5) Left shift (<<)
a = 5
print(a << 2)
```

```python
a = 5
print(a <<1)
```

# 6. Right shift (>>)

```python
# 5.6) Right shift (>>)
a = 10
print(a >> 1)
```

```python
a = 100
print(a >> 2)
```
