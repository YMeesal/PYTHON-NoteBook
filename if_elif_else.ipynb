# if elif else

# 1. simple if-else statement

```python
# 8.1) simple if-else statement
a = 5
if a > 10:
  print("a is greater than 10")
else:
  print("a is less than 10 ")
```

# 2. if-elif-else statement with multiple conditions

```python
# 8.2) if-elif-else statement with multiple conditions
a = 5
if a > 10:
  print("a is greater than 10")
elif a == 5:
  print("a is equal to 5")
else:
  print("a is less than 5 ")
```

# 3. Nested if-else statements

```python
from ast import ExtSlice
# 8.3) Nested if-else statements
a = 5
if a > 10:
  if a > 20:
    print("a is greater than 20")
  else:
    print("a is greater than 10")
else:
  print("a is less than or equal to 10")
```

# 4. checking if a number is positive or negative or zero

```python
# 8.4) checking if a number is positive or negative or zero
a = 5
if a > 0:
  print("a is positive")
elif a < 0:
  print("a is negative")
else:
  print("a is zero")
```

# 10. checking if a year is a leap year or not

```python
# 8.10) checking if a year is a leap year or not
year = 2022
if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
  print("year is a leap year")
else:
  print("year is not a leap year")
```
