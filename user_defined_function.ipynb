# 5. function to calculate the factorial of a number


---

5x4x3x2x1 = 120

---



```python
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)
```

```python
result = factorial(5)
print(result)
```

# 6. function to convert celsius to fahrenheit

```python
def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit
```

```python
result = celsius_to_fahrenheit(25)
print(result)
```

# 7. function to check if a string is palindrome or not

```python
def is_palindrome(s):
  s = s.lower()
  s = ''.join(e for e in s if e.isalnum())
  return s == s[::-1]
```

```python
result = is_palindrome("madam")
print(result)
```

```python
result = is_palindrome("hello")
print(result)
```

```python
result = is_palindrome("racecar")
print(result)
```

# 10. function to generate a fibonacci sequence

```python
def fibonacci(n):
  a = 0
  b = 1
  for i in range(n):
    print(a)
    a = b
    b = a + b
```

```python
fibonacci(10)
```

# 11. function  to check if a number is prime or not

```python
def is_prime(n):
  if n <= 1:
    return False
  for i in range(2, int(n** 0.5) + 1 ):
    if n % i == 0:
      return False
  return True
```
