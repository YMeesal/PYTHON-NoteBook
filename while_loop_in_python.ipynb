# **while loop in python**

# 1. simple while loop

```python
# 9.1) simple while loop
a = 0
while a < 5:
  print(a)
  a += 1
```

# 2. while loop with a conditional statement

```python
# 9.2) while loop with a conditional statement
a = 10
while a > 0:
  print(a)
  a -= 1
```

# 3. while loop with a break statement

```python
# 9.3) while loop with a break statement
a = 0
while a < 10:
  print(a)
  if a == 5:
    break
  a += 1
```

# 4. while loop with a continue statement

```python
# 9.4) while loop with a continue statement
a = 0
while a < 5:
  a += 1
  if a == 3:
    continue
  print(a)
```

# 5. while loop with an else clause

```python
# 9.5) while loop with an else clause
a = 0
while a < 5:
  print(a)
  a += 1
else:
  print("loop finished")
```

# 6. using while loop to iterate over a list

```python
# 9.6) using while loop to iterate over a list
fruits = ["apple", "banana", "cherry"]
a = 0
while a < len(fruits):
  print(fruits[a])
  a += 1
```

# 7. using while loop to iterate over a string

```python
# 9.7) using while loop to iterate over a string
s = "hello"
i = 0
while i < len(s):
  print(s[i])
  i +=1
```

# 8. using while loop to calculate the sum of numbers

```python
# 9.8) using while loop to calculate the sum of numbers
sum = 0
i = 1
while i <= 5:
  sum += i
  i += 1
print("the sum is", sum)
```

# 9. using while loop to calculate the factorial of a number

```python
# 9.9) using while loop to calculate the factorial of a number
n = 5
fact = 1
i = 1
while i <= n:
  fact *= i
  i +=1
print("the factorial of", n, "is", fact)
# 5x4x3x2x1 = 120
```

# 10. using while loop to check if a number is prime

```python
# 9.10. using while loop to check if a number is prime
n = 11
i = 2

while i < n:
  if n % i == 0:
    print(n, "is not a prime number")
    break
  i += 1
else:
  print(n, "is a prime number")
```

# 11. using while loop to generate a fibonacci sequence

```python
# 9.11) using while loop to generate a fibonacci sequence
a = 0
b = 1
i = 0
while i < 10:
  print(a)
  next_fib = a + b
  a = b
  b = next_fib
  i += 1
```
