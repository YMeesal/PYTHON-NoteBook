# PRINT STATEMENTS IN PYTHON

```python
# simple print
print("hello world")
```

```python
# print with variables
a = 10
print(a)
```

```python
# print with formatting
print("value of a is {}".format(a))
```

```python
# print with formatting
name = "vishal"
print("my name is {}".format(name))
```

```python
# print with multiple lines
print('hello \n world')
```

```python
# print with tab
print('hello \t world')
```

```python
# print with quotes
print("hello \"world\"")
```

```python
# print with escape characters
print("hello \\world")
```

```python
# print with raw strings
print(r"hello \world")
```

```python
# print with unicode
print("hello \u0021")
```

```python
# print with expressions
print(f"value of a is {a}")
```
