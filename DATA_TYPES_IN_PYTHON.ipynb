# **DATA TYPES IN PYTHON**

# Integer (int)

```python
# 6.1) Integer (int)
a = 5
print(a)
print(type(a))
```

# Float (float)

```python
# 6.2) Float (float)
a = 10.5
print(a)
print(type(a))
```

# Complex (complex)

```python
# 6.3) Complex (complex)
a = 10+5j
print(a)
print(type(a))
```

# Boolean (bool)

```python
# 6.4) Boolean (bool)
a = True
print(a)
print(type(a))
```

# String (str)

```python
# 6.5) String (str)
name = "meesal"
print(name)
print(type(name))
```

# List (list)

```python
# 6.6) List (list)
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(type(fruits))
```

# Tuple (tuple)

```python
# 6.7) Tuple (tuple)
Numbers = (1, 2, 3, 4, 5)
print(Numbers)
print(type(Numbers))
```

# Dictionary (dict)

```python
# 6.8) Dictionary (dict)
person = {"name":"meesal","age":23}
print(person)
print(type(person))
```

# Set (set)

```python
# 6.9) Set (set)
colors = {'red','green','blue'}
print(colors)
print(type(colors))
```

# None Type (nonetype)

```python
# 6.10) None Type (nonetype)
a = None
print(a)
print(type(a))
```

# Bytes (bytes)

```python
# 6.11)  Bytes (bytes)
data = b'hello'
print(data)
print(type(data))
```

# Range (range)


```python
# 6.12)  Range (range)
numbers = range(1,7)
print(numbers)
print(type(numbers))
```
