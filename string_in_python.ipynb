# **strings in python**

# 1. creating a string

```python
# 13.1) creating a string
s = 'i am learning python'
print(s)
print(type(s))
```

# 2. string concatenation

```python
# 13.3) string concatenation
s1 = "hello"
s2 = "world"
s3 = s1 + " " + s2
print(s3)
```

# 3. string indexing

```python
# 13.4) string indexing
s = 'hello world'
print(s[0])
```

# 4. slicing

```python
# 13.5) slicing
s = 'hello world'
print(s[0:5])
print(s[6:])
```

# 5.  Length

```python
# 13.6) Length
s = 'hello'
print(len(s))
```

# 7. methods

```python
# 13.7) methods
s = 'hello world'
print(s.upper())
print(s.capitalize())
print(s.title())
```

# 7. formatting

```python
# 13.8) formatting
name = "meesal"
age = 30
print(f"my name is {name} and i am {age} year old.")
```

# 8. splitting

```python
# 13.9) splitting
s = 'apple, banana, cherry, mango,graph'
fruits = s.split(",")
print(fruits)
```

# 9. joining

```python
# 13.10) joining
fruits = ["apple","banana","cherry","mango","graph"]
s = ",".join(fruits)
print(s)
```

# 10. searching

```python
# 13.11) searching
fruits = ["apple","banana","cherry","mango","graph"]
print(fruits.index("banana"))
```
